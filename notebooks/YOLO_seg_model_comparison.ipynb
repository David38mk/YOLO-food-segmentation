```python
import os
import cv2
import matplotlib.pyplot as plt
import numpy as np
from ultralytics import YOLO

# Load SEGMENTATION model
model = YOLO("../models/yolov8l-seg.pt")  # or yolov8n-seg.pt etc.

image_folder = r"..\\data\\FoodID_Dataset"
images = sorted([os.path.join(image_folder, img) for img in os.listdir(image_folder)
                 if img.lower().endswith((".jpg", ".png", ".jpeg"))])

for idx, img_path in enumerate(images):
    results = model(img_path)

    img = cv2.imread(img_path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    result = results[0]

    # ------------------------------------------------
    # Draw polygon masks (OUTLINE ONLY)
    # ------------------------------------------------
    if result.masks is not None:
        for mask in result.masks.xy:
            pts = mask.astype(np.int32)
            cv2.polylines(img, [pts], True, (255, 255, 0), 2)  # Yellow polygon outline

    # ------------------------------------------------
    # Draw bounding boxes
    # ------------------------------------------------
    for box in result.boxes:
        x1, y1, x2, y2 = box.xyxy[0]
        cls = int(box.cls[0])
        conf = float(box.conf[0])
        label = f"{model.names[cls]} {conf:.2f}"

        cv2.rectangle(img, (int(x1), int(y1)), (int(x2), int(y2)), (0, 128, 255), 2)
        cv2.putText(
            img, label, (int(x1), int(y1)-5),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,128,255), 2
        )

    # ------------------------------------------------
    # Show result
    # ------------------------------------------------
    plt.figure(figsize=(8,6))
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Segmentation (Polygon Only) – Image {idx}")
    plt.show()
```
